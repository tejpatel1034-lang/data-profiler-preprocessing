```python
import numpy as np

print("NumPy Version:", np.__version__)
```

```python
scalar = np.array(5)

print("Scalar:", scalar)
print("Dimensions:", scalar.ndim)
print("Shape:", scalar.shape)
```

```python
vector = np.array([10, 20, 30, 40])

print("Vector:", vector)
print("Dimensions:", vector.ndim)
print("Shape:", vector.shape)
```

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Matrix:\n", matrix)
print("Dimensions:", matrix.ndim)
print("Shape:", matrix.shape)
```

```python
tensor = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("Tensor:\n", tensor)
print("Dimensions:", tensor.ndim)
print("Shape:", tensor.shape)
```

```python
# Array of zeros
zeros = np.zeros((3, 3))

# Array of ones
ones = np.ones((3, 3))

# Sequence of numbers
numbers = np.arange(1, 11)

# Random numbers
random_values = np.random.rand(3, 3)

print("Zeros:\n", zeros)
print("Ones:\n", ones)
print("Numbers:", numbers)
print("Random Values:\n", random_values)
```

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = np.dot(A, B)

print("Matrix Multiplication:\n", result)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

result = arr + 5

print(result)
```
