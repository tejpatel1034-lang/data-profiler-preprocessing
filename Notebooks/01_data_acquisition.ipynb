##### Practical implementation 03

```python
import sqlite3
import os
```

```python
# Database path
db_path = "../database/customers.db"

# Create database folder if it doesn't exist
os.makedirs("../database", exist_ok=True)

# Connect to SQLite database
connection = sqlite3.connect(db_path)

print("Database connected successfully!")
```

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS customers (
    CustomerID INTEGER PRIMARY KEY,
    Name TEXT,
    Age INTEGER,
    Gender TEXT,
    Income REAL,
    Purchases INTEGER,
    Churn INTEGER
);
"""

connection.execute(create_table_query)

print("Customer table created successfully!")
```

```python
customer_data = [
    (101, "Tej", 23, "Male", 70000, 12, 0),
    (102, "Vandita", 25, "Female", 55000, 8, 1),
    (103, "Diya", 23, "Female", 65000, 20, 0),
    (104, "Tisa", 27, "Female", 48000, 15, 0),
    (105, "Nidhi", 35, "female", 45000, 5, 1)
]
```

```python
insert_query = """
INSERT OR REPLACE INTO customers
(CustomerID, Name, Age, Gender, Income, Purchases, Churn)
VALUES (?, ?, ?, ?, ?, ?, ?)
"""
```

```python
connection.executemany(insert_query, customer_data)
connection.commit()
```

```python
print("Customer records inserted successfully!")
```

```python
query = "SELECT * FROM customers"

sql_df = pd.read_sql_query(query, connection)

sql_df
```

```python
query = """
SELECT CustomerID, Name, Income
FROM customers
WHERE Income > 50000
"""

high_income_customers = pd.read_sql_query(query, connection)

high_income_customers
```

```python
query = "SELECT COUNT(*) AS Total_Customers FROM customers"

result = pd.read_sql_query(query, connection)

print(result)
```

```python
connection.close()

print("Database connection closed successfully!")
```

```python
import requests
```

```python
# API URL
api_url = "https://jsonplaceholder.typicode.com/users"
```

```python
# Send GET request
response = requests.get(api_url, timeout=30)
```

```python
# Check response status
print("Status Code:", response.status_code)
```

```python
if response.status_code == 200:
    api_data = response.json()
    print("API data fetched successfully!")
else:
    print("API request failed:", response.status_code)
```

```python
print(json.dumps(api_data[:2], indent=4))
```

```python
# Convert JSON into DataFrame
api_df = pd.json_normalize(api_data)
```

```python
# Display first five records
api_df.head()
```

```python
selected_columns = [
    "id",
    "name",
    "username",
    "email",
    "phone",
    "website",
    "address.city",
    "company.name"
]
```

```python
api_customers = api_df[selected_columns]
```
