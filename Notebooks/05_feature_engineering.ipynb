```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import (
    LabelEncoder,
    OneHotEncoder,
    StandardScaler,
    MinMaxScaler
)
```

```python
df = pd.read_csv(
    "../Data/Processed/cleaned_customers.csv"
)

print("Dataset Shape:", df.shape)

df.head()
```

```python
if {"PurchaseAmount", "PurchaseFrequency"}.issubset(df.columns):

    df["AveragePurchaseValue"] = (
        df["PurchaseAmount"] /
        df["PurchaseFrequency"].replace(0, np.nan)
    )

    print("New feature created!")
```

```python
if "Age" in df.columns:

    df["AgeGroup"] = pd.cut(
        df["Age"],
        bins=[0, 25, 35, 50, 100],
        labels=[
            "Young",
            "Adult",
            "Middle Aged",
            "Senior"
        ]
    )

    print("Age groups created!")
```

```python
df_encoded = df.copy()

if "Gender" in df_encoded.columns:

    le = LabelEncoder()

    df_encoded["Gender_Encoded"] = le.fit_transform(
        df_encoded["Gender"].astype(str)
    )

    print("Label Encoding Completed!")

df_encoded.head()
```

```python
if "Membership" in df.columns:

    df_onehot = pd.get_dummies(
        df,
        columns=["Membership"],
        dtype=int
    )

    display(df_onehot.head())
```

```python
numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

scaler = StandardScaler()

df_standardized = df.copy()

if numeric_columns:
    df_standardized[numeric_columns] = scaler.fit_transform(
        df[numeric_columns]
    )

df_standardized.head()
```

```python
minmax_scaler = MinMaxScaler()

df_normalized = df.copy()

if numeric_columns:
    df_normalized[numeric_columns] = minmax_scaler.fit_transform(
        df[numeric_columns]
    )

df_normalized.head()
```

```python
print("Available Features:")

for column in df.columns:
    print(column)
```

```python
selected_features = df.select_dtypes(
    include=np.number
)

display(selected_features.head())
```
